```python
# Workshop 2: Grover bounded search
# Verify empirical query count matches the theoretical bound.
import math
from poc.src.grover.amplifier import GroverAmplifier
from poc.src.grover.oracle import BoundedOracle
```

```python
N = 128
candidates = list(range(N))
oracle_fn = lambda c: c == 42
amp = GroverAmplifier(oracle=oracle_fn, backend='classical')
result = amp.amplify(candidates)
theory = GroverAmplifier.theoretical_iterations(N, 1)
print('theory_iters:', theory, 'actual_iters:', result.iterations, 'winner:', result.winner)
```

```python
# Classical baseline: brute-force query count equals N for worst case.
brute = BoundedOracle(oracle_fn).evaluate(candidates)
print('brute_queries:', brute.query_count, 'grover_queries:', result.query_count)
```
