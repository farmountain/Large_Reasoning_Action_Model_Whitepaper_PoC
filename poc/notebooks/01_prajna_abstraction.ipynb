```python
# Workshop 1: Prajna Abstraction
# Maps a raw symbolic problem into a bounded candidate set
# indexed by mined invariants.
import sympy as sp
from poc.src.prajna.abstraction import PrajnaAbstractor
from poc.src.prajna.invariant_miner import InvariantMiner
```

```python
x, y = sp.symbols('x y')
raw = 'x**2 + 2*x*y + y**2'
ab = PrajnaAbstractor(lam=0.5, miner=InvariantMiner())
out = ab.abstract(raw, [x, y], lambda inv: tuple((i, 'factored') for i in inv))
print('invariants:', out.invariants)
print('candidate_space_size:', len(out.candidate_space))
```

```python
# Compression ratio sweep
import random, sympy as sp
from poc.src.prajna.abstraction import PrajnaAbstractor
x = sp.symbols('x')
ab = PrajnaAbstractor()
for raw_size in [16, 64, 256]:
    out = ab.abstract('x**3 + 1', [x], lambda inv: tuple(range(min(raw_size, 8))))
    print(f'raw={raw_size} -> compressed={len(out.candidate_space)}')
```
