```python
# Workshop 3: Tier-3 validation (formal + causal + circuit)
from poc.src.tier3.validator import Tier3Validator
from poc.src.tier3.causal_checker import CausalChecker, CausalQuery
from poc.src.tier3.circuit_probe import CircuitProbe
```

```python
cc = CausalChecker(use_dowhy=False)
cp = CircuitProbe()
v = Tier3Validator(
    formal=lambda c: 'proof' in str(c).lower(),
    causal=lambda c: cc.is_identifiable(CausalQuery(treatment='t', outcome='y')),
    circuit=lambda c: cp.score([1, 2, 3], [1, 2, 3]).faithful,
)
print('ok:', v.validate('proof of lemma 1').accepted)
print('fail:', v.validate('wrong').accepted)
```
