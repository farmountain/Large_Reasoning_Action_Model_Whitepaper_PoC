```python
# Workshop 5: Teacher-student distillation
import numpy as np
from poc.src.teacher_student.distill import DistillationTrainer, Teacher
```

```python
teachers = [
    Teacher('LLM', lambda s: np.array([0.6, 0.2, 0.2]), weight=0.5),
    Teacher('LAM', lambda s: np.array([0.2, 0.6, 0.2]), weight=0.3),
    Teacher('SLM', lambda s: np.array([0.2, 0.2, 0.6]), weight=0.2),
]
t = DistillationTrainer(teachers=teachers, n_actions=3, lr=0.1)
q_ref = np.array([0.4, 0.4, 0.2])
state = np.array([0.0, 0.0, 0.0])
totals = [t.step(state, q_ref)['total'] for _ in range(50)]
print('loss reduction:', totals[0], '->', totals[-1])
```
