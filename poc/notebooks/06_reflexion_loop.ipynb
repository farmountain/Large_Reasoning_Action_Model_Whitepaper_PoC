```python
# Workshop 6: Reflexion-enhanced outer loop
from poc.src.teacher_student.reflexion import ReflexionLoop, ReflexionMemory
```

```python
mem = ReflexionMemory(decay=0.9)
outcomes = iter([False, False, True, True, True, True])
loop = ReflexionLoop(
    validator=lambda c: next(outcomes),
    reflect_fn=lambda t, c: f'reject {c}',
    memory=mem,
)
for step in range(6):
    r = loop.step({'step': step}, f'cand{step}')
    print(step, r)
print('memory replay:', mem.replay())
```
