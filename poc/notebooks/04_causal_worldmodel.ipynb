```python
# Workshop 4: Causal reasoning + world-model distillation
import numpy as np
from poc.src.world_model.distill_world import Transition, WorldModelDistiller
from poc.src.world_model.do_attention import DoAttention
```

```python
rng = np.random.default_rng(0)
data = []
for _ in range(64):
    s = rng.standard_normal(3)
    a = rng.standard_normal(2)
    sn = s + 0.1 * np.array([a[0], a[1], 0.0])
    data.append(Transition(s, a, float(np.sum(s)), sn))
wm = WorldModelDistiller(state_dim=3, action_dim=2)
losses = wm.fit(data, epochs=40, lr=0.01)
print('loss start->end:', losses[0], '->', losses[-1])
```

```python
att = DoAttention(d_k=3, beta=0.5)
Q = np.eye(3); K = np.eye(3); V = np.eye(3)
print('no do:', att(Q, K, V).round(3))
print('do(x0):', att(Q, K, V, do_x=np.array([1.0, 0.0, 0.0])).round(3))
```
